### clase agente blackJack 🕵️‍♀️

```python
import gymnasium as gym
import numpy as np
from torch import nn
import torch

class Agente(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear_relu_stack = nn.Sequential(
            nn.Linear(3, 64),  ## 
            nn.ReLU(),
            nn.Linear(64, 64),
            nn.ReLU(),
            nn.Linear(64, 2),
        )

    def forward(self, x):
        logits = self.linear_relu_stack(x)
        return logits
    
def preprocess_state(state):
    # Convertimos la tupla (suma, carta_crupier, as) en un tensor flotante
    return torch.tensor([state[0], state[1], float(state[2])], dtype=torch.float32)
```

### inicializacion del agente 🎬

```python
# Training hyperparameters
learning_rate = 0.01        # How fast to learn (higher = faster but less stable)
n_episodes = 1000       # Number of hands to practice

# Create environment and agent
env = gym.make("Blackjack-v1", sab=False)
env = gym.wrappers.RecordEpisodeStatistics(env, buffer_length=n_episodes)

agent = Agente()
optimizer = torch.optim.Adam(agent.parameters(), lr=learning_rate)
loss_fn = nn.MSELoss()
```

### entranamiento del agente 🏋️

```python
from tqdm import tqdm  # Progress bar
import random
import os
import gc

for episode in tqdm(range(n_episodes)):
    # Start a new hand
    obs, info = env.reset()
    state = preprocess_state(obs)
    done = False
    total_reward = 0

    # Play one complete hand
    while not done:
        q_values = agent(state)
        action = torch.argmax(q_values).item() if random.random() > 0.1 else env.action_space.sample()

        # Take action and observe result
        next_obs, reward, terminated, truncated, info = env.step(action)

        # Move to next state
        done = terminated or truncated

        target = reward + (0.95 * torch.max(agent(preprocess_state(next_obs))).item() if not done else 0) # type: ignore
        current_q = q_values[action]

        loss = loss_fn(current_q, torch.tensor(target))
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        state = preprocess_state(next_obs)

env.close()
```
